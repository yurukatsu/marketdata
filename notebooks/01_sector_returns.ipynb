# MSCI EM セクター別リターン

`configs/universe.yaml` の `sectors` の終値（USD）を取得し、期間リターンを比較する。

```python
from marketdata import LSEGClient, alias_map, load_universe, plot_period_returns

universe = load_universe()
sectors = universe["sectors"]
client = LSEGClient()
```

```python
START, END = "2026-08-01", "2026-10-02"

df = client.get_history(
    universe=[s.ticker for s in sectors],
    fields=["TR.PriceClose(Curn=USD)"],
    start=START,
    end=END,
)
df = df.rename(columns=alias_map(sectors)).ffill()
df.tail()
```

```python
returns, fig = plot_period_returns(df, START, END, title="MSCI EM セクター別リターン")
fig.show()
```

```python
client.close()
```
