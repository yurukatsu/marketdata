# ベンチマーク構成銘柄の SEDOL → RIC 変換

入力: `data/reference/bm.csv`（sedol, isin, name, cap, weight）

```python
import pandas as pd

from marketdata import REFERENCE_DIR, LSEGClient

bm = pd.read_csv(REFERENCE_DIR / "bm.csv", dtype={"sedol": str}, encoding="utf-8-sig")
client = LSEGClient()
```

```python
df = client.get_data(
    universe=[f"{s}@SEDOL" for s in bm["sedol"]],
    fields=["TR.RIC", "TR.CommonName", "TR.ExchangeName"],
)
df
```

```python
client.close()
```
