# 国別株価指数リターン

`marketdata history countries` で保存した `data/raw/countries.csv` を読み込んで可視化する。

```python
import pandas as pd

from marketdata import RAW_DIR, plot_period_returns

df = pd.read_csv(RAW_DIR / "countries.csv", index_col="date", parse_dates=True)
df.tail()
```

```python
returns, fig = plot_period_returns(
    df, "2026-01-01", "2026-09-30", title="国別株価指数リターン", xaxis_title="国"
)
fig.show()
```
